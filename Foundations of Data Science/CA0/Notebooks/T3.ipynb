# Drug Safety Test

The dataset comes from a randomized controlled drug trial conducted by a medical group
and shared by Vanderbilt University Department of Biostatistics. The study evaluates drug
safety by comparing a Drug and Placebo group while tracking adverse effects, vital signs,
and lab measures .You will conduct hypothesis testing using t-tests to determine statistical
differences between groups.
Columns Explanation:

- age: Age of the participant
- sex: Gender of the participant (male or female).
- trx: Treatment group:
    - "Drug" → Received the actual drug.
    - "Placebo" → Received a placebo (control group).
- week: Week number in the study.
- wbc: White blood cell count (WBC) measurement.
- rbc: Red blood cell count (RBC) measurement.
- adverse_effects: Presence of adverse effects (Yes or No).
- num_effects: Number of adverse effects experienced by the participant.

> We are going to check if there is a significant difference between drug group and placebo group statics using `t-test`

## Part 6

### First step

We filter the data set bu `trx` type and now we have two new data frame one for `Drug` and one for `placebo`.

Here is our hypotheses:
- H₀: There is no significant difference
- H₁: There is a significant difference

We are going to use independent t-test because we use this t-test for independent samples when we want to compare the means of two independent groups or samples. We want to know if there is a significant difference between these means.


```python
drug_group = df[df["trx"] == "Drug"]
placebo_group = df[df["trx"] == "Placebo"]

print(drug_group.head())
print()
print(placebo_group.head())
```

### Second Step

basically we can calculate the `t-value` of t-test with a simple formula that requires sample means and SE s the formula is shown below:

![image.png](./images/T3-1.jpg)


Then using `t-value` and `degree of freedom` we can calculate `p-value` which is usually done by tables and calculators
and then using `t-value` and `degree of freedom` we can find `p-value` (using tables and calculators)

here we use scipy built-in method `ttest_ind` that does this all. 

> The `p-value` is the probability of getting the observed results (or something more extreme) if the null hypothesis is true.
we defina a `significant level (a)` and if p-value is less than that it means we reject H₀ otherwise we can't reject H₀

in the below part you can see the result of t-test for different significant levels.

The significance level (a) is the threshold we use to decide whether a result is statistically significant. also it shows the probility of type 1 error.

```python
alpha_levels = [0.05, 0.1]
metrics = ["wbc", "rbc", "num_effects", "adverse_effects"]

for metric in metrics:
    t_stat, p_value = ttest_ind(drug_group[metric], placebo_group[metric], equal_var=False)
    print(f"{metric}:") 
    print(f"t-stat = {t_stat:.4f}, p-value = {p_value:.4f}")

    for alpha in alpha_levels:
        if p_value < alpha:
            print(f"Significant difference found with alpha = {alpha} (p < {alpha})")
        else:
            print(f"No significant difference with alpha = {alpha} ({alpha} <= p)")
    print()
print()
```

- What is the alternative argument and what did you choose for each metric, why?

    The alternative argument in `ttest_ind()` specifies the type of hypothesis test: `two-sided` , `less`, `greater` that two sided is the defualt option means alternative hypothese is both higher and lower. For this study, we typically use `two-sided` because we want to check if there is any difference (higher or lower) between the Drug and Placebo groups for each metric.

    However, if we had a prior expectation (for example "The drug should reduce adverse effects"), we might use "less" for num_effects and adverse_effects. but we don't have such a expectation.

- What is the equal_var argument and what does it do?

    The `equal_var` argument controls whether we assume that the two groups have equal variances.

    If equal_var=True, it assumes both groups have the same variance, which makes the test more stable but may be incorrect. If equal_var=False, it does not assume equal variances which is more reliable when the variances differ.
    Since real-world medical data often has different variances between groups, we usually set equal_var=False for better accuracy. we also have a different formula for equal variance that you can read more about it in [here](https://datatab.net/tutorial/unpaired-t-test)
